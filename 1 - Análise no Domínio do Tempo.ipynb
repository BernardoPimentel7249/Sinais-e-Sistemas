        Prefácio

    <p style="text-indent: 2em;">
        <i>Sistemas lineares diferenciais</i>, uma classe de sistemas LCIT (Lineares, Contínuos e Invariantes no Tempo), serão o foco de análise desta seção. Nesse contexto, entradas x(t) estão relacionadas a saídas y(t) por meio de EDOs da forma
</div>


$$
\frac{d^N y}{dt^N} 
+ a_1 \frac{d^{N-1}y}{dt^{N-1}} 
+ \cdots 
+ a_{N-1} \frac{dy}{dt} 
+ a_N y(t) 
= b_{N-M} \frac{d^M x}{dt^M} 
+ b_{N-M+1} \frac{d^{M-1}x}{dt^{M-1}} 
+ \cdots 
+ b_{N-1} \frac{dx}{dt} 
+ b_N x(t).
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Para tornar a notação mais concisa e para facilitar operações posteriores, definimos o operador diferencial <i>p</i> :
</div>

$$
p = \frac{d}{dt}, 
\qquad 
p^k y(t) = \frac{d^k y}{dt^k},
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    tal que a EDO genéria acima pode ser reescrita como
</div>

$$
\sum_{n=0}^{N} a_n \, p^{\,N-n} \, y(t)
=
\sum_{m=0}^{M} b_{\,M-m} \, p^{\,m} \, x(t)
$$

<div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
(a<sub>0</sub> = 1 e N ≤ M)
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    e o lado esquerdo e o lado direito podem ser expressados na forma polinomial Q(<i>p</i>) e P(<i>p</i>) respectivamente. Encerrando essa introdução, nota-se que por consequência da propriedade de superposição de entradas em sistemas LCIT(linearidade), a resposta total pode ser decomposta na soma de dois componentes: a resposta de entrada nula (y<sub>z</sub>) e a resposta de estado nulo(y<sub>s</sub>). Sendo assim, vamos estudar a obtenção de cada uma dessas parcelas da saída total e alguns atributos do sistema no domínio temporal. 
</div>

$$
y(t) = y_{z}(t) + y_{s}(t).
$$

<div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
(t ≥ 0)
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h2>
        1 - Resposta de Entrada Nula
    </h2>
    <p style="text-indent: 2em;">
        A resposta de entrada nula é a solução da EDO regente do sistema, sujeita a condições iniciais e na ausência de forças externas (x(t) ≡ 0). Com efeito, se a equação for de ordem ≥ 2 basta resolver a equação propondo soluções proporcionais a uma exponencial natural, cujos expoentes serão raízes do polinômio característico do sistema e cujos coeficientes serão determinados pelas condições iniciais (ou seja, resolvemos um PVI de uma EDO homogênea, como no cálculo 2). Para EDOs de ordem 1, podemos usar o método de separação de variáveis.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        1.1 - Exemplo Com EDO de Primeira Ordem
    </h4>
    <p style="text-indent: 2em;">
        Vamos começar com uma EDO de primeira ordem
</div>

$$
3\frac{dy}{dt} 
+ 6y(t)
= x(t),
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    sujeita à condição inicial y(0<sup>-</sup>) = 1/4. Para determinar a resposta de entrada nula, fazemos x(t) = 0.
</div>

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
t, C = sp.symbols('t,C')
y = sp.Function('y_z')(t)
# Aqui escrevemos a eq. como se ambos os lados fossem divididos por 3, para que a0 seja 1 e isolamos y(t):
eq = sp.Eq(sp.diff(y) , -2*y)
display(eq)
```

```python
# Dividindo os dois lados por y:
eq_div = sp.Eq((1/y) * sp.diff(y) , -2)
display(eq_div)
```

```python
# "Passamos" dt para o lado direito. dy/dt não é uma fração então, a rigor, essa operação não é válida, porém é um atalho conveniente
# que nos leva ao mesmo resultado que o "caminho correto". Feito isso, integramos o lado esquerdo com respeito a y e o direito 
# com respeito a t.
eq_int = sp.Eq(sp.integrate(1/y, y), sp.integrate(-2, t)+C)
display(eq_int)
sol = sp.solve(eq_int,y)
display(sp.Eq(y, sol[0]))
# Aqui consideramos e^C = C por simplicidade -> exp(C-2t) = C exp(-2t)
```

```python
# Por fim, resta aplicar a condição incial:
display(sp.Eq(y, sol[0].subs(sp.exp(C), 1/4)), t >= 0)
```

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        1.2 - Exemplo Com EDO de Segunda Ordem
    </h4>
    <p style="text-indent: 2em;">
        Agora, vamos considerar uma EDO dos slides da aula 04 sobre análise no domínio do tempo:
</div>

$$(p^{2} + 3p + 2)\,y(t) = 0, \quad y(0) = 0, \quad \dot{y}(0) = -5$$

```python
# Vamos supor uma solução na forma exp(rt). Pra obter os valores de r, aplicamos Q em r -> Q(r): 
r, y = sp.symbols('r, y_z')
eq_caracteristica = sp.Equality(r**2 + 3*r + 2, 0)
display(eq_caracteristica)

r = sp.solve(eq_caracteristica)
display('raízes:', r)
```

```python
# Logo, a solução é da forma A*exp(-2t) + B*exp(-t).
# Sistema linear para determinar A e B:
a = np.array([[1, 1], [-2,-1]])
b = np.array([0,-5])
x = np.linalg.solve(a, b)
A = x[0]
B = x[1]
sol = A * sp.exp(-2 * t) + B * sp.exp(-t)
display(sp.Eq(y, sol), t >= 0)
```

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <p style="text-indent: 2em;">
    Recordamos que o espaço vetorial que contém a solução dessas EDOs possui dimensão N, onde N é a ordem da EDO. Portanto, espera-se que ao tratarmos da resposta de entrada nula de um sistema, obtenhamos N soluções linearmente independentes para formar uma base desse espaço.
    <p style="text-indent: 2em;">
    Se houverem raízes 'r' repetidas, obtemos uma solução exp(r) e multiplicamos cada termo exp(r) subsequente por potências inteiras de t, começando por t<sup>1</sup>, depois t<sup>2</sup>, assim por diante. Exemplo: uma raiz λ se repete 4 vezes em um polinômio característico, tal que haverá termos A exp(λ) + B t exp(λ) + C t<sup>2</sup>exp(λ) + D t<sup>3</sup> exp(λ) na solução da EDO.
    <p style="text-indent: 2em;">
    Para sistemas reais, com quais nós estamos lidando, a resposta y(t) também deve ser real. Então mesmo que haja raízes complexas, a solução pode ser reduzida a uma função real. Isso ocorre pois as raízes complexas surgem em pares conjugados, de forma que se tivermos r<sub>1</sub> = a + jb e r<sub>2</sub> = a - jb, o termo da solução A exp(r<sub>1</sub>) + B exp(r<sub>2</sub>) se reduz a C exp(at) cos(bt + θ), onde θ decorre do fato de A e B serem coeficientes complexos também conjugados.
</div>

    <h2>
        2 - Resposta de Estado Nulo
    </h2>
    <p style="text-indent: 2em;">
        Anteriormente, abordamos a solução do sistema omitindo agitações externas e considerando todas suas "energias internas". Agora, vamos direcionar nosso foco à análise do sistema quando não há energias internas, mas há a ação de uma força exterior. Na prática, isso significa que as condições iniciais serão nulas e x(t) ≢ 0. A seguir, vamos introduzir o impulso unitário, o conceito de resposta ao impulso e uma nova operação matemática (a convolução), entendendo como esses elementos interagem entre si para produzir a resposta de estado nulo.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        2.1 - O Delta de Dirac
    </h4>
    <p style="text-indent: 2em;">
        Uma força altamente concentrada em forma de sinal, é de grande interesse na análise de um problema físico. Afinal, a partir da descrição de um fenômeno pontual, espera-se que possamos inferir a caracterização de uma situação maior ao aglomerar tais forças localizadas (por exemplo, quando integram-se cargas pontuais sob um volume para obter o campo eletrostático de uma região). Por isso, ao longo dos anos, eruditas de diversos campos se depararam com o desejo de expressar um ‘impulso unitário’ matematicamente.
    <p style="text-indent: 2em;">
        Um nome recorrente na teoria de sinais e sistemas (e na engenharia elétrica como um todo) é o de Oliver Heaviside, matemático e engenheiro eletricista em grande parte responsável pelo estopim do desenvolvimento do ferramental matemático usado até hoje para analisar sistemas. Heaviside foi um dos primeiros a utilizar uma forma do impulso unitário em seus trabalhos sobre física, porém sem grande comoção da comunidade científica e até com certo desprezo dos matemáticos. Foi apenas durante a era da mecânica quântica no século XX, onde, muito provavelmente inspirado por Heaviside, Paul Dirac apresentou e popularizou o conceito da função delta (δ).
    <center><img src="imagens/Paul Dirac.jpg" width="250" height="312" alt="aproxima" ></center>
    <div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
        (Paul Dirac)
    </div>
    <p style="text-indent: 2em;">
        Em sinais e sistemas, é possível obter a saída de estado nulo (y<sub>s</sub>) de um sistema através do delta por meio da função de resposta ao impulso unitário (verdade seja dita, é possível obter y<sub>s</sub> usando várias outras funções, contudo a formulação da resposta ao impulso é um tanto conveniente. Veremos o porquê disso ao longo do curso). Mas afinal, qual é a descrição analítica do delta? Antes de ver isso, é preciso conscientizar-se sobre sua natureza. O delta de dirac é uma chamada ‘função generalizada’ ou ‘distribuição’, uma classe de objetos matemáticos, formalizados apenas no século anterior, que estendem o conceito de funções típicas do cálculo.
    <p style="text-indent: 2em;">
        Uma particularidade das funções generalizadas é o fato de serem “... definidas por seu efeito em outras funções em vez de seus valores em todo instante de tempo.”(LATHI, 2007, p. 95). Por conseguinte, o delta porta um sentido operacional mais do que um sentido numérico em nossas análises. Enfim, almeja-se um impulso concentrado na região x = ∈, onde ∈ é um número suficientemente próximo, mas não igual, a zero:
</div>

$$
\delta(t) = 0 \quad para \quad t \neq \epsilon
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Além disso, a ‘força total’ do impulso unitário deve ser 1. Logo: 
</div>

$$
\int_{-\infty}^{\infty}\delta(t)dt = 1 .
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <p style="text-indent: 2em;">
        Munidos dessa informação, pode-se mostrar que o delta obedece uma importante propriedade de amostragem e uma relação com o degrau unitário u(t):
</div>

$$
\int_{-\infty}^{\infty}f(t)\delta(t - \tau)dt = f(\tau) \quad e \quad \frac{du}{dt} = \delta(t).
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <p style="text-indent: 2em;">
        Para nossas finalidades, a informação acima é suficiente, encerrando a discussão sobre o delta. A derivação rigorosa do delta vai além do escopo desse notebook e pode ser feita de várias maneiras. Uma pintura do delta é feita tomando limites de sequências de certas funções elementares contínuas (alguns tipos de funções racionais ou portas retangulares por exemplo), que dá ao delta uma aparência de ‘espinho’ infinitamente alto e de largura infinitesimal.

</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        2.2 - O Que é a Resposta ao Impulso?
    </h4>
    <p style="text-indent: 2em;">
        Dado um sinal x(t), podemos aproximá-lo por uma sequência de pulsos retangulares:
</div>

$$
x_p(t) = \sum_{n=-\infty}^{\infty} x(n \Delta \tau)\,\mathrm{rect}\!\left(\frac{t - n\Delta \tau}{\Delta \tau}\right), \quad x_p(t){\approx} x(t).
$$

<center><img src="imagens/aproxima.jpg" width="500" height="250" alt="aproxima" ></center>
<div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 14px;">
        Fonte: 'Sinais e Sistemas Lineares - 2ª ed. - B.P. Lathi'
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    No sentido de uma distribuição, a aproximação converge para x(t) se tomarmos o limite de Δ𝜏 tendendo a 0:
</div>

$$
\lim_{\Delta \tau \to 0} x_p(t) 
= \int_{-\infty}^{\infty} x(\tau)\,\delta(t-\tau)\,d\tau 
= x(t).
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <p style="text-indent: 2em;">
        Agora, vamos introduzir o conceito da resposta ao impulso h(t). Seja h(t) a saída de um sistema linear diferencial, oriundo da transformação aplicada pelo sistema sobre a distribuição delta de Dirac: <i>T</i> {δ(t)}.
</div>

<center><img src="imagens/transforma.png" width="500" height="200" alt="transforma" ></center>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Em seguida, explorando o fato do sistema ser linear, escrevemos a saída geral y(t) = <i>τ</i> {x(t)} como
</div>

$$
y(t)
= T \{\int_{-\infty}^{\infty} x(\tau)\,\delta(t-\tau)\,d\tau \}
= \int_{-\infty}^{\infty} x(\tau)\,T\{\delta(t-\tau)\}d\tau .
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    É possível reescrever y(t) conforme o lado esquerdo visto que, a transformação é aplicada em t e não 𝜏, e que o sistema é invariante ao deslocamento. Focando no termo <i>T</i> {δ(t - 𝜏)}, percebe-se que ele é justamente a resposta ao impulso h(t) deslocado no tempo. A integral do lado esquerdo é recorrente na engenharia, na física e na matemática, merecendo então um nome especial: integral de convolução!
</div>

$$
x_1(t) * x_2(t)
= \int_{-\infty}^{\infty} x_1(\tau)\,x_2(t-\tau)d\tau
$$

<div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
(integral de convolução. '*' denota convolução)
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Em conclusão, isso significa que conhecendo a resposta ao impulso h(t) de um sistema, podemos obter a resposta do sistema a qualquer entrada x(t),   convoluindo x(t) com h(t). Muito bem, mas como determinamos h(t)?
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        2.3 - Determinação da Resposta ao Impulso
    </h4>
    <p style="text-indent: 2em;">
        Para abordar a obtenção da resposta ao impulso, vamos partir de um exemplo e do fato de a resposta ao impulso ser dada por h(t) = <i>β</i> δ(t) + modos característicos (veremos o que cada um desses termos significa):
    <p style="text-indent: 2em;">
        Ex: Obter a resposta ao impulso h(t) do sistema descrito por
</div>

$$
\frac{d^2y}{dt^2}
+ 2\frac{dy}{dt}
+ y(t)
= \frac{d^2x}{dt^2}
+ \frac{dx}{dt}
+x(t).
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Definição: definimos o "sistema elementar" como Q(p)y(t) = x(t), tal que os modos naturais (y<sub>n</sub>) do sistema são soluções do sistema elementar com x(t) = δ(t). Para tal finalidade, resolvemos o sistema Q(p)y(t) = 0, representando o efeito do impulso por meio das condições iniciais para t < 0<sup>-</sup> e t > 0<sup>+</sup>.
</div>

```python
t = sp.symbols('t')
y_natural = sp.Function('y_n')(t)
sistema_elementar = y_natural.diff(t,t) + 2*y_natural.diff(t) + y_natural
display('sistema elementar:', sp.Eq(sistema_elementar, sp.DiracDelta(t)))
modos_naturais = sp.dsolve(sistema_elementar, y_natural)
display(modos_naturais)
```

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <p style="text-indent: 2em;">
    Como há apenas um δ(t) no lado direito, é necessário que haja um salto de descontinuidade na derivada de ordem N - 1 dos modos
    naturais y<sub>n</sub>(t) no ponto t = 0. Isso implica que as condições iniciais serão 
</div>

$$
y^{(N-1)}(0^+) = 1 \quad ,\quad  y^k(0^+) = y^{k-1}(0^+) = ... = y'(0^+) = y(0^+) = 0 \quad \quad (k \in \mathbb{Z} / k < N - 1, t > 0^+),
$$
$$
y^{(N-1)}(0^+) = y^k(0^-) = y^{k-1}(0^+) = ... = y'(0^+) = y(0^+) = 0 \quad \quad (t < 0^-).
$$

```python
# No caso do exemplo atual: y_n(0-) = 0 ; y_n'(0-) = 0     t < 0-
#                           y_n(0+) = 0 ; y_n'(0+) = 1     t > 0+

# Sistema linear para determinar C_1 e C_2:
a = np.array([[1, 0], [-1,1]])
b = np.array([0,1])
x = np.linalg.solve(a, b)
C_1 = x[0]
C_2 = x[1]
# Como y_n(0-), multiplicamos a solução por pelo degrau u(t):
u = sp.Function('u')(t)
modos_naturais = (C_1 + C_2 * t) * sp.exp(-t) * u
display('modos naturais:', modos_naturais)
```

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
<p style="text-indent: 2em;">
    Os modos naturais são a resposta impulsional do sistema elementar correspondente e são o primeiro passo para obter h(t). Ou seja, sendo os modos naturais dados por Q(p)y<sub>n</sub> = δ, se aplicarmos o operador P(p) aos dois lados, obtemos
    Q(p)P(p)y<sub>n</sub> = P(p)δ. Isso é muito semelhante à resposta ao impulso do sistema original Q(p)h = P(p)δ, porém o termo P(p)y<sub>n</sub> está no lugar de h(t).
    Assim, concluímos que h(t) = P(p)y<sub>n</sub>.
</p>

<p style="text-indent: 2em;">
    No nosso caso, aplicando P(p) a y<sub>n</sub>, chegamos a h(t) = (t exp(-t) - exp(-t)) u(t).
    Se derivarmos essa função descontínua uma vez, o resultado possuirá um termo δ(t), e ao derivar h(t) pela segunda vez, obtemos um termo
    proporcional à derivada primeira do impulso. Porém, ao analisar o lado esquerdo do sistema original e substituindo x por δ, percebe-se
    que deveria haver uma derivada segunda do impulso (d²x/dt²). Portanto, para sistemas nos quais a ordem de Q(p) é a mesma de P(p),
    introduzimos o termo <i>β</i> δ(t) mencionado anteriormente a h(t). Feito isso, h(t) irá satisfazer a resposta ao impulso.
</p>

</div>

```python
# Fazendo o casamento de impulsos no nosso exemplo: β = 1.
resposta_impulso = sp.DiracDelta(t) + (t*sp.exp(-t) - sp.exp(-t)) * u
h = sp.Function('h')(t)
display('resposta ao impulso:', sp.Eq(h, resposta_impulso))
```

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        2.4 - Convolução
    </h4>
    <p style="text-indent: 2em;">
        Intuitivamente, é razoável pensar na convolução como a combinação da área de duas funções quando invertemos uma delas e a "varremos" sobre a outra. É aconselhável analisar bem a expressão da convolução para entender melhor essa afirmação. Para determinar a resposta de estado nulo, vamos fazer a seguinte convolução: 
    </p>
</div>

$$
y_z(t)
= x(t) * h(t)
= \int_{-\infty}^{\infty} x(\tau)\,h(t-\tau)d\tau.
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Por exemplo: Considere um sistema cuja entrada é x(t) = exp(-t)u(t) e cuja resposta ao impulso é h(t) = exp(-2t)u(t). Se fizermos a operação x(t) * h(t), ao visualizar o gráfico de h(t) invertido no tempo indo ao encontro de x(t), percebe-se que o produto entre as duas funções só deixa de ser nulo no intervalo 0 ≤ 𝜏 ≤ t. Portanto, a integral pode ser escrita como:
</div>

$$
x(t) * h(t)
= \int_{0}^{t} e^{-\tau}\,e^{-2(t-\tau)}d\tau.
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    Resolvendo isso, temos que x(t) * h(t) = (exp(-t) - exp(-2t))u(t).
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    O código abaixo fornece uma representação visual dessa integral de convolução:
</div>


```python
# Adaptado de https://discourse.matplotlib.org/t/convolution-integral-export-as-animation/20444 
# Essa célula é parruda, então ela pode demorar alguns minutos para rodar.
import scipy.integrate
import matplotlib.animation as animation
from IPython.display import HTML

def showConvolution(t0,f1, f2):
    convolution = np.zeros(len(t))
    for n, t_ in enumerate(t):
        prod = lambda tau: f1(tau) * f2(t_-tau)
        convolution[n] = scipy.integrate.simpson(prod(t), t)

    f_shift = lambda t: f2(t0-t)
    prod = lambda tau: f1(tau) * f2(t0-tau)

    plt.gcf().clear()
    plt.subplot(211)
    plt.gca().set_ymargin(0.05)
    plt.plot(t, f1(t), label=r'$x(\tau)$')
    plt.plot(t, f_shift(t), label=r'$h(t_0-\tau)$')
    plt.fill(t, prod(t), color='r', alpha=0.5, edgecolor='black',hatch='//')
    plt.plot(t, prod(t), 'r-', label=r'$x(\tau)h(t_0-\tau)$')
    plt.grid(True); plt.xlabel(r'$\tau$'); plt.ylabel(r'$x(\tau)$')
    plt.legend(fontsize=10)
    plt.text(-4, 0.6, '$t_0=%.2f$' % t0, bbox=dict(fc='white')) 

    plt.subplot(212)
    plt.gca().set_ymargin(0.05)
    plt.plot(t, convolution, label='$(x*h)(t)$')

    current_value = scipy.integrate.simpson(prod(t), t)
    plt.plot(t0, current_value, 'ro')
    plt.grid(True); plt.xlabel('$t$'); plt.ylabel('$(x*h)(t)$')
    plt.legend(fontsize=10)

Fs = 50 # Taxa de amostragem
T = 6 # Período de interesse [-a, a]
t = np.arange(-T, T, 1/Fs) # Discretização do tempo
f1 = lambda t: (t>0) * np.exp(-t) * 1.02 # fator 1.02 adicionado para melhorar a vizualização
f2 = lambda t: (t>0) * np.exp(-2*t)
t0 = np.arange(-2.0,6.0, 0.05)
fig = plt.figure(figsize=(8,4))
anim = animation.FuncAnimation(fig, showConvolution, frames=t0, fargs=(f1,f2), interval=80)

HTML(anim.to_jshtml())
```

    <h2>
        3 - Propriedades do Sistema
    </h2>
    <p style="text-indent: 2em;">
        A linearidade e a invariância temporal de um sistema, garantem a existência de uma função de resposta ao impulso. Além disso propriedades, podemos caracterizar um sistema em tempo contínuo pela memória, causalidade e estabilidade, entre outras classificações.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        3.1 - Memória
    </h4>
    <p style="text-indent: 2em;">
        <b>Se um sistema LCIT possuir uma resposta ao impulso que é um múltiplo escalar do próprio delta de Dirac (h(t) = aδ(t)), ele é dito sem memória</b>. Isso significa que para qualquer instante de tempo, a saída é linearmente proporcional à entrada nesse mesmo instante. Em termos de circuitos elétricos, um sistema sem memória pode ser realizado por um resistor, onde o potencial nas duas pontas diferem apenas em amplitude.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        3.2 - Causalidade
    </h4>
    <p style="text-indent: 2em;">
        <b>A causalidade de um sistema é sempre garantida quando a resposta ao impulso for nula para instantes de tempo menores que zero (h(t) = 0 ∀ t < 0)</b>. Uma alternativa de pensar sobre a causalidade é: "Um sistema é dito Causal se a sua saída em qualquer tempo depender, somente, de valores da entrada no instante presente e nos passados; pode depender, também, de valores passados da própria saída, que são as condições iniciais armazenadas nos elementos aramzenadores de energia ou informação, quando eles existirem no sistema" <a href="https://www.maxwell.vrac.puc-rio.br/22023/sistema_causal.html">(https://www.maxwell.vrac.puc-rio.br/22023/sistema_causal.html)</a>.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        3.3 - BIBO-Estabilidade
    </h4>
    <p style="text-indent: 2em;">
        BIBO é uma sigla para <i>Bounded Input Bounded Output</i>. Em outras palavras, um sistema BIBO estável é aquele cuja saída não diverge para valores infinitos se a entrada for limitada. Sendo um circuito RLC com uma fonte independente um sistema, por exemplo, não seria esperado que a tensão em qualquer terminal divirja e atinja amplitudes muito mais altas que a da fonte. <b>Um sistema é dito BIBO estável se sua função de resposta ao impulso for absolutamente integrável</b>.
</div>

$$
\int_{-\infty}^{\infty}|h(t)|dt < \infty
$$

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        3.4 - Estabilidade Assintótica
    </h4>
    <p style="text-indent: 2em;">
        Existem três tipos de sistemas no sentido de estabilidades assintótica:
        <ul>
            <li><b>Assintoticamente Estáveis:</b> possuem todas as raízes características no semi-plano esquerdo.</li>
            <li><b>Assintoticamente Instáveis:</b> possuem pelo menos uma raiz característica no semi-plano direito e/ou possuem raízes repetidas no eixo imaginário.</li>
            <li><b>Marginalmente Estáveis:</b> não possuem raízes características no semi-plano direito, porém possuem raízes características não repetidas no eixo imaginário.</li>
        </ul>
    <center><img src="imagens/estabilidade.png" width="800" height="350" alt="transforma" ></center>
</div>
<div style="text-align: right; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 14px;">
        Adaptado de 'Sinais e Sistemas Lineares - 2ª ed. - B.P. Lathi' pg. 195.
</div>
<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    A estabilidade BIBO é determinada por uma entrada externa, enquanto a estabilidade assintótica é uma descrição interna do sistema que depende das condições iniciais.
</div>

<div style="text-align: left; font-family: 'Cambria Math', 'STIX Math', 'Latin Modern Math', serif; font-size: 20px;">
    <h4>
        4.1 - Como Usar a Função resolver_sistema_LCIT( ):
    </h4>
    <p style="text-indent: 2em;">
        <h5>
        Parâmetros essenciais:
        </h5>
        <ul>
            <li>Q: Objeto lista que representa o operador Q(p) do sistema. O primeiro ítem na lista (Q[0]) representa o coeficiente da derivada de maior ordem, o segundo item o da derivada de segunda maior ordem e assim por diante, até o termo y(t).</li>
            <li>x: A entrada do sistema, escrita em termos de funções do Sympy. Um domínio de t ≥ 0 é subentendido para todo o sistema, então não é necessário multiplicar a entrada pelo degrau unitário (a não ser que a entrada seja uma constante. Ex: x = 2 * sp.Heaviside(t)).</li>
            <li>conds: Objeto lista que representa as condições iniciais do sistema para a resposta de entrada nula, começando por y(0) = a (conds[0]), depois y'(0) = b (conds[1]), e assim por diante.</li>
        </ul>
    <p style="text-indent: 2em;">
        Parâmetros opcionais:
        <ul>
            <li>P: O operador P(p). Seu valor padrão é P(1), então se ele não for especificado, a entrada da EDO será apenas x(t). Essa variável é uma lista cuja estrutura funciona como a da variável Q.</li>
            <li>T: Representa o intervalo de visualização das funções (0 < t < T). O valor padrão é 6.</li>
            <li>deslocamentos_dirac: No caso específico em que sua entrada P(p) x(t) gerar impulsos deslocados (fora da origem), os instantes em que os impulsos deslocados aparecem devem ser passados a essa variável. Ex: δ(t -2) -> deslocamentos_dirac = [2] (dos parâmetros opcionais, este talves seja o mais específico e com menor chance de requerir modificações pelo usuário).</li>
        </ul>
        Ressalta-se que essa função foi projetada com sistemas <b>causais</b> de primeira e segunda ordem em mente, visto que os sistemas de interesse prático na vida real são, geralmente, de primeira e segunda ordem. Isso se deve principalmente a dois fatos: a maioria dos sistemas da natureza ser de segunda ou primeira ordem (sistema massa-mola, circuitos RCL, pêndulo físico, ondas eletromagnéticas, segunda lei de Newton) e sistemas de ordem superior a 2 podem ser decompostos em sistemas de ordens menores (veja mais sobre isso no notebook 4). Apesar disso, existem fenômenos modelados por equações de ordem superior: "Equações de terceira ordem surgem ao modelar ondas em meios dispersivos (ondas de água ou plasma por exemplo). Equações de quarta ordem aparecem em elasticidade, em particular mecânica de placas e vigas, e em processamento de imagens. Equações de ordem igual a ou superior a 5 são muito raras (Traduzido de 'Introduction to Partial Differential Equations; Peter J. Olver')."
